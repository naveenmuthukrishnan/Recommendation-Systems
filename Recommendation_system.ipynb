```python
new_pivot = pd.pivot_table(data = data , columns = 'title', index = 'user_id', values = 'rating').fillna(0)
new_pivot
```

```python
new_simi_score = cosine_similarity(new_pivot.T)
items_collab = pd.DataFrame(new_simi_score)
items_collab.columns = data['title'].unique()
items_collab.index = data['title'].unique()
items_collab
```

```python
def ibcf(movie_name, top_movies = 5):
  if movie_name not in items_collab.columns:
    return 'Movie not found'

  scores = items_collab[movie_name].sort_values(ascending = False)[1:top_movies+1]

  return scores.index.tolist()
```

##Its an Item-Item Based Collabrative Filtering Recommendation design

##Popularity based

```python
def popularity_contents(top_movies:int):
  list_of_movies = data.groupby(by = 'title').agg({'rating':['mean','count']})
  list_of_movies.columns = ['average rating','count']
  list_of_movies = list_of_movies[list_of_movies['count'] > 50]
  list_of_movies = list_of_movies.sort_values(by = 'average rating', ascending = False)
  return list_of_movies.head(top_movies).index.tolist()

popularity_contents(top_movies=5)
```

##Trending based

```python
def trending_contents(top_movies:int):
  filtered_data = data[data['timestamp'] > '1998-01-01']
  list_of_movies = filtered_data.groupby(by = 'title').agg({'rating':['mean','count']})
  list_of_movies.columns = ['average rating','count']
  list_of_movies = list_of_movies[list_of_movies['count'] > 50]
  list_of_movies = list_of_movies.sort_values(by = 'average rating', ascending = False)
  return list_of_movies.head(top_movies).index.tolist()

trending_contents(top_movies=5)
```
